# İstatistik ve Axis

Toplam, ortalama, max gibi hesapları satır satır ya da sütun sütun yapmayı ve eksik verilerle (NaN) nasıl çalışacağımızı öğrenicez.

```python
import numpy as np
```

```python
# satırlar: mağazalar, sütunlar: günler
sales = np.array([[5, 3, 8],
                  [2, 7, 4]])
sales
```

```python
sales.sum()
```

```python
sales.sum(axis=0)
```

```python
sales.sum(axis=1)
```

```python
sales.mean()
```

```python
sales.max(), sales.argmax()
```

```python
sales.argmax(axis=1)
```

```python
sales.std().round(2)
```

```python
temps = np.array([21.0, np.nan, 23.0])
temps.mean()
```

```python
np.nanmean(temps)
```
