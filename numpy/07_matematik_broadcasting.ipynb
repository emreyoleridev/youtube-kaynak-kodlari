# Matematiksel İşlemler ve Broadcasting

Array'lerle eleman eleman matematiksel işlemler yapmayı ve farklı boyuttaki array'lerle döngü yazmadan nasıl işlem yapabildiğimizi göricez.

```python
import numpy as np
```

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
a + b
```

```python
a * b
```

```python
b / a
```

```python
a ** 2
```

```python
np.sqrt(b).round(2)
```

```python
prices = np.array([100, 250, 80])
prices * 0.9
```

```python
table = np.array([[1, 2, 3],
                  [4, 5, 6]])
table + np.array([10, 20, 30])
```

```python
table + np.array([[100], [200]])
```

```python
np.array([1, 2]) + np.array([1, 2, 3])
```
