# Birleştirme ve Kopyalama

Array'leri birleştirip bölmeyi ve view ile copy arasındaki o kritik farkı göricez.

```python
import numpy as np
```

```python
a = np.array([[1, 2],
              [3, 4]])
b = np.array([[5, 6]])
np.vstack([a, b])
```

```python
np.hstack([a, np.array([[9], [9]])])
```

```python
np.split(np.arange(6), 3)
```

```python
original = np.array([1, 2, 3, 4])
part = original[:2]
part[0] = 99
original
```

```python
original = np.array([1, 2, 3, 4])
safe = original[:2].copy()
safe[0] = 99
original
```
