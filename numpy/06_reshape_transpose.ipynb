# Reshape ve Transpose

Array'in şeklini değiştirmeyi ve satırlarla sütunların yerini değiştirmeyi göricez.

```python
import numpy as np
```

```python
numbers = np.arange(12)
numbers
```

```python
table = numbers.reshape(3, 4)
table
```

```python
numbers.reshape(2, -1)
```

```python
numbers.reshape(5, 3)
```

```python
table.T
```

```python
table.ravel()
```

```python
column = np.array([1, 2, 3]).reshape(-1, 1)
column
```
