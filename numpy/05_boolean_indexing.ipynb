# Boolean Indexing

Bir koşul yazıp array'in içinden işimize yarayan elemanları süzmeyi ve filtrelediğimiz elemanların değerini değiştirmeyi öğrenicez.

```python
import numpy as np
```

```python
scores = np.array([45, 82, 67, 91, 38, 74])
passed = scores >= 50
passed
```

```python
scores[passed]
```

```python
passed.sum()
```

```python
scores[(scores >= 60) & (scores < 80)]
```

```python
scores[[0, 2, 4]]
```

```python
scores[scores < 50] = 50
scores
```
