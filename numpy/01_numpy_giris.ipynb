# NumPy'a Giriş

Python listesiyle NumPy array'i arasındaki farkı görücez ve NumPy'ın neden bu kadar hızlı olduğunu anlıycaz.

```python
import numpy as np
np.__version__
```

```python
[1, 2, 3] * 2
```

```python
np.array([1, 2, 3]) * 2
```

```python
numbers = list(range(1_000_000))
%timeit [n * 2 for n in numbers]
```

```python
numbers_np = np.arange(1_000_000)
%timeit numbers_np * 2
```
