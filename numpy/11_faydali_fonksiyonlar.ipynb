# Faydalı Fonksiyonlar

Günlük hayatta en çok kullandığımız where, sort, unique ve clip fonksiyonlarını göricez.

```python
import numpy as np
scores = np.array([45, 82, 67, 91, 38, 74])
```

```python
np.where(scores >= 50, "geçti", "kaldı")
```

```python
np.sort(scores)
```

```python
np.argsort(scores)
```

```python
np.clip(scores, 50, 90)
```

```python
colors = np.array(["red", "blue", "red", "green", "blue", "red"])
np.unique(colors, return_counts=True)
```
