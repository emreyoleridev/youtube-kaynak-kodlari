Matris çarpımı yapıcaz, denklem sistemi çözücez ve ters matrise bakıcaz.

```python
import numpy as np
```

```python
A = np.array([[2, 1],
              [1, 3]])
A * A
```

```python
A @ A
```

```python
x = np.array([1, 2])
A @ x
```

```python
# 2x +  y = 5
#  x + 3y = 10
b = np.array([5, 10])
np.linalg.solve(A, b)
```

```python
np.linalg.inv(A).round(2)
```
