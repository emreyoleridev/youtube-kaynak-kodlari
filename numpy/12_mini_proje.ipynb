# Mini Proje: Not Analizi

Bir sınıfın notlarını alıp öğrendiğimiz her şeyle baştan sona analiz edicez.

```python
import numpy as np
```

```python
names = np.array(["Ali", "Ayşe", "Can", "Deniz", "Elif"])
# satırlar: öğrenciler, sütunlar: vize, ödev, final
grades = np.array([[60, 80, 55],
                   [90, 85, 95],
                   [40, 70, 45],
                   [75, 60, 80],
                   [85, 95, 70]])
grades.shape
```

```python
grades.mean(axis=0)
```

```python
weights = np.array([0.3, 0.2, 0.5])
final_scores = grades @ weights
final_scores
```

```python
best = final_scores.argmax()
names[best], final_scores[best]
```

```python
names[final_scores >= 60]
```

```python
names[np.argsort(final_scores)[::-1]]
```

```python
normalized = (grades - grades.mean(axis=0)) / grades.std(axis=0)
normalized.round(2)
```
