```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Data from the previous parts
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/emreyoleridev/youtube-kaynak-kodlari/main/data/netflix_clean.csv")
movie_df = df[df["type"] == "Movie"]
ratings = ["TV-MA", "TV-14", "R", "PG-13"]
groups = [movie_df[movie_df["rating"] == r]["duration_num"] for r in ratings]
x = np.linspace(0, 10, 100)
n = np.arange(11)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 3))
ax1.plot(x, np.sin(x))
ax2.plot(x, np.cos(x))
plt.title("sin")
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 3))
ax1.plot(x, np.sin(x))
ax2.plot(x, np.cos(x))
ax1.set_title("sin")
ax2.set_title("cos")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(n, n ** 2)
plt.show()
plt.savefig("empty.png")
```
