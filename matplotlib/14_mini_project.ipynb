# 14 · Mini Project: Netflix Dashboard

Four charts, one page: a Netflix overview dashboard from the clean dataset.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Data from the previous parts
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/emreyoleridev/youtube-kaynak-kodlari/main/data/netflix_clean.csv")
counts = df["type"].value_counts()
minutes = df[df["type"] == "Movie"]["duration_num"]
top = df["main_country"].value_counts().drop("Unknown").head(8).sort_values()
```

```python
per_year = df["year_added"].value_counts().sort_index().loc[2015:2020]
per_year
```

```python
fig, axs = plt.subplots(2, 2, figsize=(11, 7), layout="constrained")
axs[0, 0].plot(per_year.index, per_year.values, marker="o")
axs[0, 0].set_title("Titles added per year")
plt.show()
```

```python
bars = axs[0, 1].bar(counts.index, counts.values, color=["tab:red", "tab:gray"])
axs[0, 1].bar_label(bars)
axs[0, 1].set_title("Movies vs TV shows")
fig
```

```python
axs[1, 0].hist(minutes, bins=40, edgecolor="white")
axs[1, 0].set_title("Movie length (minutes)")
fig
```

```python
axs[1, 1].barh(top.index, top.values)
axs[1, 1].set_title("Top countries")
fig.suptitle("Netflix in 4 charts", fontsize=16, fontweight="bold")
fig
```

```python
fig.savefig("netflix_dashboard.png", dpi=200, bbox_inches="tight")
```
