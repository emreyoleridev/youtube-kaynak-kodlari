# 03 · Figure & Axes

The object-oriented style: fig, ax = plt.subplots(), and why we do everything through ax.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 10, 100)
```

```python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x))
plt.show()
```

```python
type(fig), type(ax)
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(x, np.sin(x))
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(800, 300, "px"))
ax.plot(x, np.sin(x))
plt.show()
```
