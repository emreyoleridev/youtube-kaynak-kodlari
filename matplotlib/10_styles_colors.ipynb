Built-in styles, accessible color cycles and colormaps.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Data from the previous parts
x = np.linspace(0, 10, 100)
rng = np.random.default_rng(42)
height = rng.normal(170, 8, 200)
weight = 0.9 * height - 85 + rng.normal(0, 6, 200)
age = rng.integers(18, 70, 200)
```

```python
plt.style.available[:8]
```

```python
with plt.style.context("ggplot"):
    fig, ax = plt.subplots(figsize=(8, 3))
    ax.plot(x, np.sin(x))
    ax.plot(x, np.cos(x))
    plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
for k in range(6):
    ax.plot(x, np.sin(x + k / 2), linewidth=2)
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 3))
ax1.scatter(height, weight, c=age, cmap="jet")
ax2.scatter(height, weight, c=age, cmap="viridis")
plt.show()
```
