# 06 · Scatter Plots

Two numbers per point: scatter, alpha, and a colormap with a colorbar.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
```

```python
rng = np.random.default_rng(42)
height = rng.normal(170, 8, 200)
weight = 0.9 * height - 85 + rng.normal(0, 6, 200)
```

```python
fig, ax = plt.subplots()
ax.scatter(height, weight)
plt.show()
```

```python
fig, ax = plt.subplots()
ax.scatter(height, weight, s=15, alpha=0.5)
plt.show()
```

```python
age = rng.integers(18, 70, 200)
fig, ax = plt.subplots()
dots = ax.scatter(height, weight, c=age, cmap="viridis")
fig.colorbar(dots, label="age")
plt.show()
```
