# 04 · Line Plots

Colors, line widths, line styles, markers and a legend.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 10, 100)
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(x, np.sin(x), color="tab:orange", linewidth=3)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(x, np.sin(x), linestyle="--")
ax.plot(x, np.cos(x), linestyle=":")
plt.show()
```

```python
n = np.arange(11)
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(n, n ** 2, marker="o")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(n, n ** 2, "r--o")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(x, np.sin(x), label="sin")
ax.plot(x, np.cos(x), label="cos")
ax.legend()
plt.show()
```
