# 05 · Titles, Labels & Ticks

Titles, axis labels, ticks, limits, grid lines, spines and annotations.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
```

```python
months = np.arange(1, 13)
temps = [6, 7, 8, 12, 17, 22, 24, 25, 21, 17, 12, 9]
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(months, temps, marker="o")
plt.show()
```

```python
ax.set_title("Istanbul · average temperature")
ax.set_xlabel("Month")
ax.set_ylabel("°C")
fig
```

```python
names = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
         "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
ax.set_xticks(months, labels=names)
ax.set_ylim(0, 30)
fig
```

```python
ax.grid(alpha=0.3)
ax.spines[["top", "right"]].set_visible(False)
fig
```

```python
ax.annotate("Hottest: 25 °C", xy=(8, 25), xytext=(9.3, 28),
            arrowprops=dict(arrowstyle="->"))
fig
```
