Showing 2D data as an image with imshow and a colorbar.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
```

```python
grid = np.arange(25).reshape(5, 5)
grid
```

```python
fig, ax = plt.subplots()
ax.imshow(grid)
plt.show()
```

```python
fig, ax = plt.subplots()
pic = ax.imshow(grid, cmap="gray")
fig.colorbar(pic)
plt.show()
```

```python
checker = np.indices((8, 8)).sum(axis=0) % 2
fig, ax = plt.subplots()
ax.imshow(checker, cmap="gray")
ax.axis("off")
plt.show()
```
