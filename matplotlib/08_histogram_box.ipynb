# 08 · Histogram & Box Plot

How values are spread out: histograms, bins and box plots.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/emreyoleridev/youtube-kaynak-kodlari/main/data/netflix_clean.csv")
```

```python
movie_df = df[df["type"] == "Movie"]
minutes = movie_df["duration_num"]
minutes.describe().round(1)
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.hist(minutes)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.hist(minutes, bins=40, edgecolor="white")
ax.axvline(minutes.mean(), color="tab:red", linestyle="--")
plt.show()
```

```python
ratings = ["TV-MA", "TV-14", "R", "PG-13"]
groups = [movie_df[movie_df["rating"] == r]["duration_num"] for r in ratings]
fig, ax = plt.subplots(figsize=(8, 3))
ax.boxplot(groups, tick_labels=ratings)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.boxplot(groups, tick_labels=ratings, orientation="horizontal")
plt.show()
```
