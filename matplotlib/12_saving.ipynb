# 12 · Saving Figures

savefig: PNG vs SVG, dpi, size in pixels and tight bounding boxes.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Data from the previous parts
months = np.arange(1, 13)
temps = [6, 7, 8, 12, 17, 22, 24, 25, 21, 17, 12, 9]
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(months, temps, marker="o")
ax.set_title("Istanbul · average temperature")
fig.savefig("temps.png")
```

```python
fig.savefig("temps_hd.png", dpi=300, bbox_inches="tight")
```

```python
fig.savefig("temps.svg")
fig.savefig("temps.pdf")
fig.savefig("temps_clear.png", transparent=True)
```

```python
import os
for name in sorted(os.listdir()):
    if name.startswith("temps"):
        print(name, os.path.getsize(name) // 1024, "KB")
```
