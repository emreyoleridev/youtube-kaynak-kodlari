# 07 · Bar Charts

Comparing categories: bar, barh, grouped bars, bar labels and a pie chart.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
```

```python
import pandas as pd
URL = "https://raw.githubusercontent.com/emreyoleridev/youtube-kaynak-kodlari/main/data/netflix_clean.csv"
df = pd.read_csv(URL)
df.shape
```

```python
counts = df["type"].value_counts()
counts
```

```python
fig, ax = plt.subplots()
ax.bar(counts.index, counts.values)
plt.show()
```

```python
fig, ax = plt.subplots()
bars = ax.bar(counts.index, counts.values, color=["tab:red", "tab:gray"])
ax.bar_label(bars)
plt.show()
```

```python
top = df["main_country"].value_counts().drop("Unknown").head(8).sort_values()
fig, ax = plt.subplots()
ax.barh(top.index, top.values)
plt.show()
```

```python
years = [2016, 2017, 2018, 2019, 2020]
movie_counts = [256, 861, 1255, 1497, 1312]
show_counts = [184, 361, 429, 656, 697]
```

```python
fig, ax = plt.subplots(figsize=(8, 3.5))
ax.grouped_bar({"Movie": movie_counts, "TV Show": show_counts}, tick_labels=years)
ax.legend()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 3.5))
ax.bar(years, movie_counts, label="Movie")
ax.bar(years, show_counts, bottom=movie_counts, label="TV Show")
ax.legend()
plt.show()
```

```python
fig, ax = plt.subplots()
pie = ax.pie(counts.values, labels=counts.index)
ax.pie_label(pie, "{frac:.0%}")
plt.show()
```
