Two lists, one line of code: our first line chart, and how plt keeps track of the current plot.

```python
# Setup: the steps from the previous parts (run this first)
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 6, 5]
plt.plot(x, y)
plt.show()
```

```python
plt.plot(y)
plt.show()
```

```python
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x))
plt.title("sin & cos")
plt.show()
```
