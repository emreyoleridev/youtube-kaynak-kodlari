```python
# Data from the previous parts
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/emreyoleridev/youtube-kaynak-kodlari/main/data/netflix_clean.csv")
counts = df["type"].value_counts()
minutes = df[df["type"] == "Movie"]["duration_num"]
x = np.linspace(0, 10, 100)
rng = np.random.default_rng(42)
height = rng.normal(170, 8, 200)
weight = 0.9 * height - 85 + rng.normal(0, 6, 200)
```

```python
fig, axs = plt.subplots(2, 2)
axs[0, 0].plot(x, np.sin(x))
axs[0, 1].plot(x, np.cos(x))
axs[1, 0].scatter(height, weight, s=5)
axs[1, 1].hist(minutes, bins=30)
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 3), sharey=True)
ax1.plot(x, np.sin(x))
ax2.plot(x, 2 * np.cos(x))
plt.show()
```

```python
fig, axs = plt.subplots(2, 2)
for ax in axs.flat:
    ax.set_title("A title")
    ax.set_xlabel("x label")
plt.show()
```

```python
fig, axs = plt.subplots(2, 2, layout="constrained")
for ax in axs.flat:
    ax.set_title("A title")
    ax.set_xlabel("x label")
plt.show()
```

```python
fig, axd = plt.subplot_mosaic([["top", "top"], ["left", "right"]], layout="constrained")
axd["top"].plot(x, np.sin(x))
axd["left"].bar(counts.index, counts.values)
axd["right"].hist(minutes, bins=30)
plt.show()
```
