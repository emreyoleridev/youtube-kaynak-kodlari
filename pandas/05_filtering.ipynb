# 05 · Filtering Rows

Keeping only the rows that match a condition: masks, & and |, isin, between, query and the new filter.

```python
# Setup: the steps from the previous parts (run this first)
import pandas as pd
pd.set_option("display.max_colwidth", 30)

URL = ("https://raw.githubusercontent.com/rfordatascience/tidytuesday/"
       "master/data/2021/2021-04-20/netflix_titles.csv")
df = pd.read_csv(URL)

df = df.drop(columns=["cast", "description"])
```

```python
is_movie = df["type"] == "Movie"
is_movie.head()
```

```python
movies = df[is_movie]
movies.shape
```

```python
df[(df["type"] == "TV Show") & (df["release_year"] >= 2020)].shape
```

```python
kids = df[df["rating"].isin(["TV-Y", "TV-Y7", "TV-G"])]
kids.shape
```

```python
df[df["release_year"].between(1990, 1999)].shape
```

```python
df.query("type == 'Movie' and release_year < 1950")[["title", "release_year"]]
```

```python
df.filter(cond=pd.col("release_year") == 1925)
```
