# 02 · Series & DataFrame

The two building blocks of pandas: a single labeled column, and a table made of them.

```python
# Setup: the steps from the previous parts (run this first)
import pandas as pd
pd.set_option("display.max_colwidth", 30)
```

```python
scores = pd.Series([9.1, 8.4, 7.9], index=["Dark", "Narcos", "Ozark"])
scores
```

```python
scores["Narcos"]
```

```python
scores.mean()
```

```python
shows = pd.DataFrame({
    "title": ["Dark", "Narcos", "Ozark"],
    "seasons": [3, 3, 4],
    "my_score": [9.1, 8.4, 7.9],
})
shows
```

```python
shows["seasons"]
```
